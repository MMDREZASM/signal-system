```

import numpy as np
import matplotlib.pyplot as plt
```

```

## 🛠️ مسئله ۱: جابجایی و معکوس‌سازی زمانی (پیوسته)

# 1. تعریف متغیر زمان t
t = np.linspace(-5, 5, 500)

# تابع پله واحد (Unit Step Function)
def u(t_array):
    return np.where(t_array >= 0, 1.0, 0.0)

# 2. تعریف سیگنال اصلی x(t) = exp(-t) * u(t)
x_t = np.exp(-t) * u(t)

# 3. تعریف سیگنال تغییر یافته y(t) = x(-t - 2)
# زمان جدید: t_prime = -t - 2
t_prime = -t - 2
y_t = np.exp(-(t_prime)) * u(t_prime)

# 4. رسم نمودار
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(t, x_t, color='blue')
plt.title('سیگنال اصلی $x(t) = e^{-t}u(t)$')
plt.xlabel('زمان (t)')
plt.ylabel('دامنه')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(t, y_t, color='red')
plt.title('سیگنال تغییر یافته $y(t) = x(-t - 2)$')
plt.xlabel('زمان (t)')
plt.ylabel('دامنه')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```

## 🛠️ مسئله ۲: مقیاس‌دهی و جابجایی زمانی (گسسته)

# 1. تعریف متغیر زمان گسسته n
n = np.arange(-5, 8)

# تابع پله واحد گسسته
def u_n(n_array, n0=0):
    return (n_array >= n0).astype(int)

# 2. تعریف سیگنال اصلی x[n] = n * (u[n] - u[n-4])
x_n = n * (u_n(n, 0) - u_n(n, 4))

# 3. تعریف سیگنال تغییر یافته y[n] = x[2n - 2]
y_n = np.zeros_like(n, dtype=float)

# محاسبه y[n] به صورت حلقه:
for i, val in enumerate(n):
    # n_prime = 2*n - 2
    n_prime = 2 * val - 2

    # اگر n_prime در محدوده سیگنال اصلی (۰ تا ۳) باشد، مقداردهی شود.
    # توجه: این رویکرد فرض می‌کند که سیگنال x[n] برای n خارج از محدوده صفر است.
    if n_prime >= 0 and n_prime < 4:
        y_n[i] = n_prime # x[n_prime] = n_prime
    else:
        y_n[i] = 0

# 4. رسم نمودار
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.stem(n, x_n, linefmt='b-', markerfmt='bo', basefmt=' ')
plt.title('سیگنال اصلی $x[n]$')
plt.xlabel('$n$')
plt.ylabel('دامنه')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.stem(n, y_n, linefmt='r-', markerfmt='ro', basefmt=' ')
plt.title('سیگنال تغییر یافته $y[n] = x[2n - 2]$')
plt.xlabel('$n$')
plt.ylabel('دامنه')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```

## 🛠️ مسئله ۳: انرژی سیگنال (پیوسته)

# 1. تعریف متغیر زمان t
t = np.linspace(-1, 3, 500)
dt = t[1] - t[0] # گام زمانی برای انتگرال‌گیری عددی

# 2. تعریف سیگنال x(t) = u(t) - u(t-2)
def u(t_array):
    return np.where(t_array >= 0, 1.0, 0.0)

x_t = u(t) - u(t - 2)

# 3. محاسبه انرژی E به صورت عددی
# E = انتگرال از |x(t)|^2
integrand = np.abs(x_t)**2
E_numerical = np.sum(integrand * dt)

print(f"انرژی تئوری: E = 2")
print(f"انرژی محاسبه شده عددی: E = {E_numerical:.4f}")

# 4. رسم نمودار
plt.figure(figsize=(8, 4))
plt.plot(t, x_t, color='green')
plt.fill_between(t, x_t, color='lightgreen', alpha=0.5)
plt.title('سیگنال $x(t)$ و محاسبه انرژی')
plt.xlabel('زمان (t)')
plt.ylabel('دامنه')
plt.grid(True)
plt.show()
```

```

## 🛠️ مسئله ۴: تجزیه به بخش زوج و فرد (گسسته)

# 1. تعریف متغیر زمان n (شامل n های منفی)
n = np.arange(-5, 6)

# 2. تعریف سیگنال اصلی x[n] = u[n]
x_n = u_n(n, 0) # u_n از مسئله ۲ تعریف شده است.

# 3. محاسبه x[-n] (معکوس زمانی x[n])
# برای محاسبه x[-n]، کافی است سیگنال اصلی را با محور عمودی معکوس کنیم.
x_minus_n = u_n(-n, 0)

# 4. محاسبه بخش زوج و فرد
x_e_n = 0.5 * (x_n + x_minus_n)
x_o_n = 0.5 * (x_n - x_minus_n)

# 5. رسم نمودارها
fig, axs = plt.subplots(3, 1, figsize=(10, 10))

# x[n] اصلی
axs[0].stem(n, x_n, linefmt='b-', markerfmt='bo', basefmt=' ')
axs[0].set_title('سیگنال اصلی $x[n] = u[n]$')
axs[0].set_xlabel('$n$')
axs[0].grid(True)

# x_e[n]
axs[1].stem(n, x_e_n, linefmt='g-', markerfmt='go', basefmt=' ')
axs[1].set_title('بخش زوج $x_e[n]$')
axs[1].set_xlabel('$n$')
axs[1].grid(True)

# x_o[n]
axs[2].stem(n, x_o_n, linefmt='r-', markerfmt='ro', basefmt=' ')
axs[2].set_title('بخش فرد $x_o[n]$')
axs[2].set_xlabel('$n$')
axs[2].grid(True)

plt.tight_layout()
plt.show()
```

```

## 🛠️ مسئله ۵: بررسی خطی بودن (Linearity)

import numpy as np

# تعریف سیستم T{x} = n * x[n]
def system_T(n, x):
    return n * x

# 1. تعریف متغیر زمان و سیگنال‌های ورودی
n = np.arange(-5, 5)
x1_n = np.sin(0.5 * n)
x2_n = np.cos(0.5 * n)

# 2. تعریف ضرایب دلخواه
a = 2.0
b = 3.0

# الف. محاسبه T{a*x1 + b*x2}
x_combined = a * x1_n + b * x2_n
y_combined = system_T(n, x_combined)

# ب. محاسبه a*T{x1} + b*T{x2}
y1 = system_T(n, x1_n)
y2 = system_T(n, x2_n)
y_expected = a * y1 + b * y2

# 3. مقایسه نتایج
difference = np.max(np.abs(y_combined - y_expected))

print(f"حداکثر اختلاف بین دو عبارت: {difference:.2e}")

if difference < 1e-9:
    print("نتیجه: سیستم عدداً خطی است (اختلاف نزدیک به صفر است).")
else:
    print("نتیجه: سیستم عدداً خطی نیست.")
```

```

## 🛠️ مسئله ۶: بررسی تغییرناپذیری با زمان (Time-Invariance)

# تعریف سیستم T{x} = n * x[n] (از مسئله ۵)
def system_T(n, x):
    return n * x

# 1. تعریف متغیر زمان و سیگنال ورودی
n = np.arange(-5, 5)
x_n = np.heaviside(n, 1) # x[n] = u[n]

# 2. تعریف میزان جابجایی
n0 = 2

# الف. محاسبه T{x[n-n₀]}
n_shifted = n - n0
x_shifted = np.heaviside(n_shifted, 1) # x[n-n₀] = u[n-2]
y_input_shifted = system_T(n, x_shifted) # n * x[n-n₀]

# ب. محاسبه y[n-n₀]
y_n = system_T(n, x_n) # n * x[n]
# برای محاسبه y[n-n₀]، باید y[n] را به اندازه n₀ جابجا کنیم.
# چون y[n] به طول n است، جابجایی در آرایه numpy کمی پیچیده است.
# اما تعریف تئوری آن ساده است: (n-n₀) * x[n-n₀]
y_output_shifted_expected = (n - n0) * x_shifted

# 3. مقایسه نتایج
difference = np.max(np.abs(y_input_shifted - y_output_shifted_expected))

print(f"حداکثر اختلاف بین T{{x[n-n₀]}} و y[n-n₀]: {difference:.2e}")

if difference < 1e-9:
    print("نتیجه: سیستم عدداً تغییرناپذیر با زمان است (اشتباه!).")
else:
    print("نتیجه: سیستم عدداً تغییرپذیر با زمان است (چون اختلاف قابل توجه است).")
```

```

## 🛠️ مسئله ۷: بررسی علیت (Causality)

# علیت را نمی‌توان به سادگی با آرایه‌های ثابت اثبات کرد، اما می‌توان نقض آن را نشان داد.

# 1. تعریف متغیر زمان t
t = np.linspace(-1, 1, 100)
t0 = 0 # زمان فعلی که خروجی را در آن محاسبه می‌کنیم

# 2. تعریف ورودی
# فرض کنید ورودی x(t) فقط در t=1.5 فعال است (در آینده t0=0)
def x_future_only(t_array):
    return np.where((t_array > 1.4) & (t_array < 1.6), 5.0, 0.0)

# 3. محاسبه خروجی در t0=0
t_at_t0 = 2*t0 + 1 # 2(0) + 1 = 1
x_required_at_t0 = x_future_only(np.array([t_at_t0]))[0]

# 4. بررسی ورودی و خروجی
if x_required_at_t0 != 0:
    print(f"در زمان t₀={t0}، خروجی به ورودی در t = {t_at_t0} وابسته است.")
    print("چون 1 > 0، سیستم به ورودی آینده وابسته است.")
    print("نتیجه: سیستم غیر علی (Noncausal) است.")
else:
    print("نتوانستیم علیت را با این مثال نقض کنیم (اما تئوری آن غیرعلی است).")
```

```

## 🛠️ مسئله ۸: بررسی پایداری (Stability)

# 1. تعریف متغیر زمان n
n = np.arange(0, 10) # 10 نمونه برای شبیه‌سازی

# 2. تعریف پاسخ ضربه h[n]
alpha = 2
h_n = alpha**n

# 3. محاسبه مجموع قدر مطلق (برای بررسی پایداری)
sum_abs_h_n = np.sum(np.abs(h_n))

print(f"مجموع قدر مطلق h[n] در n=0 تا 9: {sum_abs_h_n:.2f}")

if alpha > 1:
    print("چون |α| > 1 است، مجموع به سمت بی‌نهایت میل می‌کند.")
    print("نتیجه: سیستم ناپایدار (Unstable) است.")
else:
    print("نتیجه: سیستم پایدار است.")

# 4. رسم نمودار (نشان دادن رشد نمایی)
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 4))
plt.stem(n, h_n, use_line_collection=True)
plt.title('پاسخ ضربه سیستم $h[n] = 2^n u[n]$ (ناپایدار)')
plt.xlabel('n')
plt.ylabel('دامنه')
plt.grid(True)
plt.show()
```

```

## 🛠️ مسئله ۹ و ۱۰: ترکیب خواص (علیت و پایداری)

import sympy as sp

# 1. تعریف متغیر نمادین (برای تحلیل تئوری)
t = sp.symbols('t')

# 2. تعریف پاسخ ضربه h(t) = exp(-|t|)
h_t = sp.exp(-sp.Abs(t))

print("الف) بررسی علیت (Causality):")
# علیت نیازمند h(t) = 0 برای t < 0
if h_t.subs(t, -1) != 0:
    print(f"h(-1) = {h_t.subs(t, -1):.4f} (مقدار غیر صفر).")
    print("نتیجه علیت: سیستم غیر علی (Noncausal) است.")
else:
    print("نتیجه علیت: سیستم علی است.")

print("\nب) بررسی پایداری (Stability):")
# پایداری نیازمند انتگرال |h(t)| dt < ∞
integral_h_t = sp.integrate(sp.Abs(h_t), (t, -sp.oo, sp.oo))

if integral_h_t.is_finite:
    print(f"مقدار انتگرال = {integral_h_t}")
    print("نتیجه پایداری: سیستم پایدار (Stable) است.")
else:
    print("نتیجه پایداری: سیستم ناپایدار است.")
```

```

## 🛠️ مسئله ۱۳ و ۱۴: پیچش پیوسته (شبیه‌سازی عددی)

import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve # استفاده از convolve برای سیگنال‌های پیوسته (تقریب)

# 1. تعریف متغیر زمان t
t_start = -1
t_end = 4
num_points = 500
t = np.linspace(t_start, t_end, num_points)
dt = t[1] - t[0]

# تابع پله واحد
def u(t_array):
    return np.where(t_array >= 0, 1.0, 0.0)

# 2. تعریف سیگنال‌ها
x_t = u(t) - u(t - 1) # پالس از ۰ تا ۱
h_t = u(t) - u(t - 2) # پالس از ۰ تا ۲

# 3. محاسبه پیچش y(t) = x(t) * h(t)
# توجه: convolve در scipy برای سیگنال‌های پیوسته (با گسسته‌سازی) استفاده می‌شود.
y_t_vals = convolve(x_t, h_t, mode='full') * dt

# 4. تعریف محدوده زمانی خروجی y(t)
# شروع y: t_start + t_start = -2
# پایان y: t_end + t_end = 8
# برای همگام‌سازی، باید یک بردار زمان جدید بسازیم.
t_y_end = t_start + t_end
t_y = np.linspace(2 * t_start, 2 * t_end, len(y_t_vals))

# 5. رسم نمودار
plt.figure(figsize=(8, 4))
plt.plot(t_y, y_t_vals, color='orange')
plt.title('نتیجه پیچش پیوسته $y(t) = x(t) * h(t)$ (شکل ذوزنقه)')
plt.xlabel('زمان (t)')
plt.ylabel('دامنه')
plt.xlim(-1, 3) # نمایش محدوده اصلی خروجی
plt.grid(True)
plt.show()
```

```

## 🛠️ مسئله ۱۸ تا ۲۰: اثر LTI بر سری فوریه

import sympy as sp
import numpy as np

# 1. تعریف متغیرهای نمادین
k, w0 = sp.symbols('k w0')
j = sp.I

# 2. تعریف پاسخ فرکانسی سیستم LTI
# H(jω) = 1 / (jω + 1)
H_jw = 1 / (j * w0 + 1)

# 3. محاسبه پاسخ فرکانسی سیستم در فرکانس‌های سری فوریه (ω = k * ω₀)
H_jkw0 = H_jw.subs(w0, k * w0)
print(f"پاسخ سیستم در فرکانس kω₀: H(j k ω₀) = {H_jkw0}")

# 4. استفاده از ضرایب ورودی (a_k) از مسئله قبل
# فرض کنید ورودی x(t) دارای ضرایب زیر باشد:
# a_k = (1 - (-1)^k) / (j * 2 * k * pi)
a_k = (1 - (-1)**k) / (j * 2 * k * sp.pi)

# در اینجا w0 = pi است
H_jkw0_sub = H_jkw0.subs(w0, sp.pi)

# 5. محاسبه ضرایب خروجی b_k = a_k * H(j k ω₀)
b_k = a_k * H_jkw0_sub

print("\n--- ضرایب خروجی b_k ---")
print(f"b_k = {sp.simplify(b_k)}")

# محاسبه عددی برای k=1 (به عنوان نمونه)
k_val = 1
b_1 = b_k.subs(k, k_val)
b_1_float = complex(b_1.evalf())
print(f"ضریب b₁ (عددی): {b_1_float:.4f}")

# ضریب DC (k=0)
b_0 = a0 * H_jw.subs(w0, 0) # a0=1/2, H(0)=1
print(f"ضریب b₀: {b_0}")
```
